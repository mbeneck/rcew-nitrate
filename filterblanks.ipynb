```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
aa500_result = pd.read_csv('/Users/beneck/Library/CloudStorage/OneDrive-NortheasternUniversity/Boise Project/Data/Reynolds Nitrate Monitoring/Nutrient Analyzer Data/20251020_Run/RCEW_NOx-PO4-NH3_20Oct25.TXT', header=14).rename(columns= {'Results 1': 'Nitrate', 'Results 2':'Phosphate', 'Results 3': 'Ammonium'})
samplelist = pd.read_excel('/Users/beneck/Library/CloudStorage/OneDrive-NortheasternUniversity/Boise Project/Data/Reynolds Nitrate Monitoring/Nutrient Analyzer Data/20251020_Run/samplelist_20251020.xlsx')
```

```python
# merge on 'Cup Number' to avoid overlapping column error
merged = samplelist.merge(aa500_result, on='Cup Number', how='inner')
merged
```

```python
fig, axes = plt.subplots(nrows=4, ncols=1, figsize=(8.5, 11), sharex=True)

sample_numbers = merged['Sample Number'].unique()
sample_types = ['Old', 'Unfiltered', 'Filtered After', 'Filtered Before']
rep_colors = ['tab:blue', 'tab:orange', 'tab:green']  # Rep 1, 2, 3

for i, sample_num in enumerate(sample_numbers):
    ax = axes[i]
    data = merged[merged['Sample Number'] == sample_num]
    bar_width = 0.2
    x = np.arange(len(sample_types))
    for j, st in enumerate(sample_types):
        nitrate_vals = data[data['Sample Type'] == st]['Nitrate'].values
        for k, val in enumerate(nitrate_vals):
            ax.bar(j + k * bar_width, val, width=bar_width, 
                   color=rep_colors[k % len(rep_colors)], 
                   label=f'Rep {k+1}' if i == 0 and j == 0 else "")
    ax.set_xticks(x + bar_width)
    ax.set_xticklabels(sample_types)
    ax.set_title(f'Sample Number: {sample_num}')
    if i == 0:
        handles, labels = ax.get_legend_handles_labels()
        by_label = dict(zip(labels, handles))
        ax.legend(by_label.values(), by_label.keys(), title='Replicate')
fig.suptitle('Nitrate')
plt.xlabel('Sample Type')
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(nrows=4, ncols=3, figsize=(24, 16), sharex=True)

sample_numbers = merged['Sample Number'].unique()
sample_types = ['Old', 'Unfiltered', 'Filtered After', 'Filtered Before']
analytes = ['Nitrate', 'Phosphate', 'Ammonium']
rep_colors = ['tab:blue', 'tab:orange', 'tab:green']  # Rep 1, 2, 3

for row, sample_num in enumerate(sample_numbers):
    data = merged[merged['Sample Number'] == sample_num]
    x = np.arange(len(sample_types))
    bar_width = 0.2
    for col, analyte in enumerate(analytes):
        ax = axes[row, col]
        for j, st in enumerate(sample_types):
            vals = data[data['Sample Type'] == st][analyte].values
            for k, val in enumerate(vals):
                ax.bar(j + k * bar_width, val, width=bar_width,
                       color=rep_colors[k % len(rep_colors)],
                       label=f'Rep {k+1}' if row == 0 and col == 0 and j == 0 else "")
        ax.set_xticks(x + bar_width)
        ax.set_xticklabels(sample_types)
        if row == 0:
            ax.set_title(analyte)
        if col == 0:
            ax.set_ylabel(f'Sample {sample_num}')
        if row == 0 and col == 0:
            handles, labels = ax.get_legend_handles_labels()
            by_label = dict(zip(labels, handles))
            ax.legend(by_label.values(), by_label.keys(), title='Replicate')

fig.suptitle('Analyte Concentrations by Sample Type and Replicate', fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.show()
```

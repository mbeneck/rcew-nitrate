## Total RME

```python
def get_filtered_sample_ids(sample_data, site_names, bottle_replicate):
    selected = sample_data[sample_data['Site Name'].isin(site_names)]
    selected = selected[selected['Bottle Replicate'] == bottle_replicate]
    selected = selected.set_index(['Sample Collected', 'Time Sample Collected'])
    sample_ids = selected['Sample ID'].to_frame()
    sample_ids.to_clipboard(index=False)
    return selected
```

```python
get_filtered_sample_ids(samples, ['RME', 'Field Blank', 'RME Snow'], 1)
```

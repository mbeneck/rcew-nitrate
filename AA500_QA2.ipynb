## Joining all Results

```python
def merge_duplicate_samples(df, sample_id_col='Sample ID', values = ['Nitrate', 'Phosphate', 'Ammonium']):

    def merge_rows(group):
        if len(group) == 1:
            return group
        else:
            for value in values:
                columns = [value + param for param in [' mean', ' std', ' err', ' QA']]
                # check if there are non problematic QA samples
                good_samps = ~(group[value + ' QA'].str.contains('BP') | group[value + ' QA'].str.contains('BBV') | group[value + ' QA'].str.contains('ISV'))
                # take samples without QA flags
                if(good_samps.sum()>0):
                    group.loc[~good_samps, columns] = np.NaN
                # set all but lowest std samples to nan
                min_std = group[value + ' std'].idxmin()
                if pd.notna(min_std):
                    group.loc[group.index != min_std, columns] = np.NaN
            
            
            merged = group.apply(lambda x: x.dropna().iloc[0] if not x.dropna().empty else np.nan)
            return pd.DataFrame([merged])

    # Group by sample_id and apply the merge function
    result = df.groupby(sample_id_col).apply(merge_rows)
    return result


def join_site_results(sitename, result_list):
    joined = pd.DataFrame()
    for result in result_list:
        df = result.get_condensed_data().reset_index().set_index('Sample ID')
        joined = pd.concat([joined, df[df['Site Name'] == sitename]])
    no_dupes= merge_duplicate_samples(joined)
    return no_dupes.reset_index().drop(columns=['level_1']).set_index('Sample Datetime').sort_index()

```

```python
rme_results = join_site_results('RME', [result_102225, result_102325, result_102725, result_102825, result_102925, result_103025, result_110725])
fieldblank_results = join_site_results('Field Blank', [result_102225, result_102325, result_102725, result_102825, result_102925, result_103025, result_110725])
rme_snow_results = join_site_results('RME Snow', [result_102225, result_102325, result_102725, result_102825, result_102925, result_103025, result_110725])
```

```python
def plot_between_bottle_bar(result_df, analyte, **kwargs):
    # Reset index to ensure 'Sample Datetime' and 'Bottle Replicate' are columns
    df = result_df.reset_index()

    # Pivot so each bottle replicate is a column, indexed by date
    pivoted = df.pivot_table(
    index='Sample Datetime',
    columns='Bottle Replicate',
    values=analyte + ' mean'
    )

    # Optionally, get errors for yerr
    yerr = df.pivot_table(
        index='Sample Datetime',
        columns='Bottle Replicate',
        values=analyte + ' err'
    )

    # Plot
    ax = pivoted.plot(kind='bar', yerr=yerr, rot=45, **kwargs)
    ax.set_ylabel(analyte+ ' mean')
    #ax.set_title(analyte + ' by Date and Bottle Replicate')
    ax.legend(title='Bottle Replicate')
    #plt.tight_layout()
    #plt.show()
    return ax
```

```python
fig, ax = plt.subplots(nrows=3, figsize=(11,11), sharex=True)

plot_between_bottle_bar(rme_results.loc['4/1/25':'5/1/25'], 'Nitrate', ax=ax[0], title='Nitrate BBV')
plot_between_bottle_bar(rme_results.loc['4/1/25':'5/1/25'], 'Phosphate', ax=ax[1], title='Phosphate BBV')
plot_between_bottle_bar(rme_results.loc['4/1/25':'5/1/25'], 'Ammonium', ax=ax[2], title='Ammonium BBV')


fig.tight_layout()
```
